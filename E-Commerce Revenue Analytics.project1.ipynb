```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv(r'C:\Users\rajat\Downloads\Superstore data analysis.csv', encoding="unicode_escape")
```

```python
df.head()
```

```python
df.info()
```

```python
df["Order Date"]=pd.to_datetime(df["Order Date"])
```

```python
df["Ship Date"]=pd.to_datetime(df["Ship Date"])
```

```python
df.columns=[col.replace(' ','_') for col in df.columns]
```

```python
df.columns
```

```python
pd.isnull(df)
```

```python
pd.isnull(df).sum()
```

```python
df.dropna()
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df['Unit_Price']=df["Sales"]/df["Quantity"]
```

```python
print("Rows:", df.shape[0])
```

```python
print("Columns:", df.shape[1])
```

```python
df["Year"] = df["Order_Date"].dt.year
df["Month"] = df["Order_Date"].dt.month
df["Month_Name"] = df["Order_Date"].dt.strftime("%b")
df["Quarter"] = "Q" + df["Order_Date"].dt.quarter.astype(str)
```

```python
df["Profit_margin%"]=df["Profit"].sum() / df["Sales"].sum()*100
```

```python
df["Shipping_days"]=(df["Order_Date"]-df["Ship_Date"]).dt.days
```

```python
df["Month_Name"] = df["Order_Date"].dt.strftime("%b")
```

```python
df[["Order_Date","Year","Month","Month_Name","Quarter","Sales","Profit","Profit_margin%","Shipping_days"]].head(10)
```

```python
df.columns
```

```python
total_orders=df["Order_ID"].nunique()
total_customers=df["Customer_ID"].nunique()
total_products=df["Product_ID"].nunique()
total_revenue=df["Sales"].sum()
total_profit=df["Profit"].sum()
profit_margin= total_profit / total_revenue * 100
average_order_value= total_revenue / total_orders

```

```python
kpis = pd.DataFrame({
    "KPI": ["Total Revenue","Total Profit","Total Orders","Total Customers",
            "Total Products","Profit Margin","Average Order Value"],
    "Value": [total_revenue,total_profit,total_orders,total_customers,
              total_products,profit_margin,average_order_value]
})
kpis
```

```python
categorywise_analysis=(
    df.groupby("Category")
    .agg(Revenue=("Sales","sum"),
        Profit=("Profit","sum"),
        Orders=("Order_ID","nunique"),
        Quantity=("Quantity","sum")
    )
    .sort_values("Revenue", ascending= False)
                      
)
categorywise_analysis.round(2)
```

```python
plt.figure(figsize=(10,5))
sns.barplot(data=categorywise_analysis, x= "Revenue", y="Orders", color="yellow")
plt.xlabel("Revenue")
plt.ylabel("Order")
plt.xticks(rotation=0)
plt.title("Revenue vs Order")
plt.show()
```

```python
monthly_sales = (
    df.groupby(df["Order_Date"].dt.to_period("M"))
      .agg(Revenue=("Sales","sum"), Profit=("Profit","sum"))
)
monthly_sales.index = monthly_sales.index.to_timestamp()
monthly_sales.head()
```

```python
monthly_sales["Month_Name"] = monthly_sales.index.strftime("%b")
plt.figure(figsize=(12,5))
sns.barplot(data=monthly_sales, x= "Month_Name", y="Revenue", color="Red")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=0)
plt.title("Revenue vs Month")
plt.show()
```

```python
customer_analysis = (
    df.groupby(["Customer_ID","Customer_Name"])
      .agg(Orders=("Order_ID","nunique"), Revenue=("Sales","sum"),
           Profit=("Profit","sum"), Quantity=("Quantity","sum"))
      .sort_values("Revenue", ascending=False)
)
customer_analysis.head(10).round(2)
```

```python
first_purchase = (
    df.groupby("Customer_ID")["Order_Date"]
      .min()
      .rename("First Purchase Date")
      .reset_index()
)

df = df.merge(first_purchase, on="Customer_ID", how="left")
df["Customer Type"] = np.where(
    df["Order_Date"] == df["First Purchase Date"],
    "New Customer",
    "Returning Customer"
)

customer_type = (
    df.groupby("Customer Type")
      .agg(Revenue=("Sales","sum"), Orders=("Order_ID","nunique"),
           Customers=("Customer_ID","nunique"))
)
customer_type.round(2)

```

```python

product_analysis = (
    df.groupby(["Product_ID","Product_Name"])
      .agg(Revenue=("Sales","sum"), Profit=("Profit","sum"),
           Quantity=("Quantity","sum"), Orders=("Order_ID","nunique"))
)

print("Top 10 products by revenue:")
display(product_analysis.sort_values("Revenue", ascending=False).head(10).round(2))

print("\nBottom 10 products by profit:")
display(product_analysis.sort_values("Profit").head(10).round(2))
```

```python
discount_analysis = (
    df.groupby("Discount")
      .agg(Revenue=("Sales","sum"), Profit=("Profit","sum"),
           Average_Profit=("Profit","mean"), Orders=("Order_ID","nunique"))
      .sort_index()
)
discount_analysis.round(2)

```

```python
print("Correlation between discount and profit:",
      round(df["Discount"].corr(df["Profit"]), 3))
```

```python
sns.scatterplot(data=discount_analysis, x="Revenue", y="Average_Profit", color="purple")
```

```python
region_analysis = (
    df.groupby("Region")
      .agg(Revenue=("Sales","sum"), Profit=("Profit","sum"),
           Orders=("Order_ID","nunique"))
      .sort_values("Revenue", ascending=False)
)
region_analysis.round(2)
```

```python
df.to_csv("superstore_cleaned.csv", index=False)
print("File saved successfully!")
print("Final shape:", df.shape)
```
